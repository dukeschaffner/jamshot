## Testing Section

Now let's test the implementation with a sample track ID.


```python
# Test with real database data
# Replace TRACK_ID with an actual track ID from your database
TRACK_ID = 120  # Update this with a real track ID

print("🔌 Testing with real database data...")
print(f"📊 Fetching track tree for track ID: {TRACK_ID}")

# Initialize data collector (uses real database)
data_collector = DataCollectionModule()

try:
    # Fetch real track data from database
    tracks = data_collector.collect_all_data(TRACK_ID)
    
    print(f"✅ Successfully fetched {len(tracks)} tracks from database")
    
    # Test frame generation with real data
    print("\n🎨 Testing frame generation with real data...")
    
    frame_generator = FrameGenerationModule()
    test_frame = frame_generator.generate_frame(tracks, "test_frame.png", playback_time=5)
    
    print(f"✅ Test frame generated: {test_frame.size}")
    print("📁 Check test_frame.png to see the output")
    
except Exception as e:
    print(f"❌ Error: {e}")
    print("\n💡 Make sure:")
    print("   1. Database connection is configured (check .env file)")
    print("   2. TRACK_ID exists in the database")
    print("   3. Track has processing_status = 'completed'")
```

```python
# Test full video generation
# This will generate a complete video file with animated waveforms

TRACK_ID = 121  # Update this with a real track ID

print("🎬 Testing full video generation...")
print(f"📊 Generating video for track ID: {TRACK_ID}")

# Initialize pipeline with local saving enabled
pipeline = VideoExportPipeline(save_locally=True, local_dir="./test_output")

try:
    # Generate full video
    result = pipeline.export_track_video(TRACK_ID)
    
    if result['success']:
        print(f"\n✅ Video generation successful!")
        print(f"📹 Video path: {result['video_path']}")
        print(f"⏱️  Duration: {result['duration']:.2f}s")
        print(f"🎞️  FPS: {result['fps']}")
        print(f"📐 Frame size: {result['frame_size']}")
        print(f"🎵 Tracks: {result['track_count']}")
    else:
        print(f"\n❌ Video generation failed: {result.get('error', 'Unknown error')}")
        
except Exception as e:
    print(f"❌ Error: {e}")
    import traceback
    traceback.print_exc()
```
